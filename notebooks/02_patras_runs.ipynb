# Patras runs

Each cell trains one model per fold and
writes `<name>_preds.csv` and `<name>_folds.csv` to `outputs/runs/patras/`.
Frames and test folds are read from `outputs/manifests/patras/`.

```python
import sys
sys.path.insert(0, "..")

import pandas as pd

from src.run import get_device, load_config, run

cfg = load_config("../configs/patras.yaml", data_root="../data/patras")
OUT = "../outputs/runs/patras"
print("device:", get_device())


def manifest(name):
    return pd.read_csv(f"../outputs/manifests/patras/{name}.csv")
```
